# Mutual Fund NAV Data Cleaning & Preprocessing Pipeline
This notebook loads raw daily NAV series, handles missing dates/non-trading days, imputes missing values, and computes daily log returns and trailing 30-day volatility.

```python
import pandas as pd
import numpy as np
import os

# Load cleaned dataset or fallback
file_path = 'data/cleaned/clean_nav.csv'
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print('Data loaded successfully. Head:')
    print(df.head())
else:
    print('File not found. Please run generate_analytics_data.py first.')
```

```python
# Data Quality Check & Missing Value Analysis
print('Missing Values Check:')
print(df.isnull().sum())

# Forward-fill missing NAV values for non-trading days
df['nav_date'] = pd.to_datetime(df['nav_date'])
df = df.sort_values(['fund_id', 'nav_date'])
df['nav_value'] = df.groupby('fund_id')['nav_value'].ffill()

print('\nData cleaning complete. Total rows processed:', len(df))
```
