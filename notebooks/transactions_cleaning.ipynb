 Day 2 - Task 2: Investor Transactions Cleaning

```python
import pandas as pd
import os

df = pd.read_csv("../data/raw/08_investor_transactions.csv")

df.head()
```

```python
df.info()
```

```python
print(df.columns)
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print(df["transaction_type"].unique())
```

```python
df["transaction_type"] = (
    df["transaction_type"]
    .str.strip()
    .str.title()
)
```

```python
print(df["transaction_type"].unique())
```

```python
invalid_amount = df[df["amount_inr"] <= 0]

print("Invalid Amount Records:", len(invalid_amount))
```

```python
print(df["kyc_status"].unique())
```

```python
print(df["transaction_date"].head())
```

```python
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"],
    format="%Y-%m-%d"
)
```

```python
os.makedirs("../data/cleaned", exist_ok=True)

df.to_csv("../data/cleaned/clean_transactions.csv", index=False)

print("Cleaned transactions dataset saved successfully!")
```
